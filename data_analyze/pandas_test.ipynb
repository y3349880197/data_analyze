```python
#数据过滤
df_4 = pd.read_csv("./data/sales.csv")
df_4

#销售数量>=10的订单数据
df_4[df_4['销售数量']>=10]

#产品类型为食品饮料或家居日用的订单数据
df_4[df_4['产品类别'].isin(['食品饮料','家居日用'])]

#单价在100-200之间的订单数据,包含边界
df_4[df_4['单价'].between(100, 200)]

#销售数量>=8,且单价>=100的订单数据
df_4[(df_4['销售数量']>=8) & (df_4['单价']>=100)]

#产品类别为家居日用/图书文具, 支付方式为支付宝/微信支付的订单数据
df_4[(df_4['产品类别'].isin(['家居日用', '图书文具'])) & (df_4['支付方式'].isin(['支付宝', '微信支付']))]
```

```python
#数据清洗
df_5 = pd.read_csv('data/films.csv')

df_5

#缺失值处理
df_5.isnull()#查看缺失值
#1.删除缺失值,并不是在原本的上面删,返回一个新的
df_5.dropna()#删除缺失值所在行
df_5.dropna(axis=1)#删除缺失值所在列

#2.填充缺失值
df_5.fillna('---')
df_5.ffill()#使用上一行数据来填充
df_5.bfill()#使用下一行数据填充

#重复值处理
df_5.duplicated()#查看重复值,所有的列的数据都一致才算重复
df_5.duplicated(subset=['电影名'])#查看重复值,指定列重复算重复

df_5.drop_duplicates(subset=['电影名'])#删除重复值

#异常值处理
df_5[df_5['评分']<0]#查看异常数据

df_5.drop(df_5[df_5['评分']<0].index)#删除异常数据
df_5['评分'] = df_5['评分'].abs()#修复异常数据

#数据格式转换
df_5['上映时间'] = df_5['上映时间'].str.replace('/', '-')
```

```python
#数据分组
df_7 = pd.read_csv('data/sales.csv', nrows=30)

#根据产品类别进行分组,统计各个类别的销售数量
df_7.groupby('产品类别')['销售数量'].count()

df_7.groupby('产品类别')['单价'].min()

df_7.groupby('产品类别')['单价'].max()

df_7.groupby('产品类别')['单价'].mean()

df_7.groupby('产品类别')['单价'].agg(['min', 'max', 'mean'])

df_7.groupby('产品类别').agg({'销售数量': 'sum', '总金额':'sum', '单价':'mean'})
```
