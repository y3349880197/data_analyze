```python
import matplotlib.pyplot as plt
import random

plt.rcParams['font.sans-serif'] = ['Arial Unicode MS']  # macOS 自带中文字体
plt.rcParams['axes.unicode_minus'] = False              # 解决负号显示成方块

#绘制折线图
x = [i for i in range(1, 25)]
y = [random.randint(10, 15) for i in x]
y_js = [random.randint(13, 18) for i in x]

plt.figure(figsize=(8, 5))#宽 高
plt.plot(x, y, label='北京') #如果没有画布就会自动创建一个,要想有画布,必需要在plot前面
plt.plot(x, y_js, label='江苏')

#设置折线图详细信息
plt.title('气温变化折线图') #标题
plt.xlabel('时间') #x轴标签
plt.ylabel('温度') #y轴标签

plt.xticks(x[1::2]) #x轴刻度
y_ticks = [i for i in range(5, 21)] #y轴刻度
plt.yticks(y_ticks)

plt.grid(linestyle='--') #绘制网格线

plt.legend(loc='lower left') #设置图例

plt.show()#显示图表
```

```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Arial Unicode MS']  # macOS 自带中文字体
plt.rcParams['axes.unicode_minus'] = False              # 解决负号显示成方块

#figure:画布对象
#axes:子图数组
figure, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 6), dpi=100) #行 列

axes_1:Axes = axes[0]#柱状图
axes_2:Axes = axes[1]#饼图

axes_1.set_title('世界石油储备', fontsize=20)
countries = ['中国', '美国', '英国', '加拿大', '沙特', '伊朗']
values = [35, 23, 18, 21, 56, 78]
axes_1.bar(countries, values, width=0.6, color='g')
axes_1.set_xlabel('国家')
axes_1.set_ylabel('是有储备量')
axes_1.grid(linestyle='--', alpha=0.3)

axes_2.set_title('世界人口比重', fontsize=20)
countries2 = ['印度', '中国', '美国', '印尼', '巴基斯坦', '尼日利亚', '巴西', '俄罗斯', '其他']
values2 = [14.51, 14.09, 3.4, 2.83, 2.51, 2.33, 2.12, 1.44, 20]
axes_2.pie(values2, labels=countries2, autopct='%1.1f%%')
axes_2.legend(loc="lower center", ncol=5, bbox_to_anchor=(0.5, -0.1))

plt.savefig('data/01.png')#保存图像

plt.show()
```
